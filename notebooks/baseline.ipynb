# 1 - Data

## Classes and Functions

#### BirdCLEFData

```python
class BirdCLEFData:
    def __init__(self, data_path: Path):
        self._data_path = data_path

        self._taxonomy_path = data_path / "taxonomy.csv"
        self._sample_submission_path = data_path / "sample_submission.csv"
        self._train_soundscapes_labels_path = data_path / "train_soundscapes_labels.csv"
        self._train_path = data_path / "train.csv"

        self._taxonomy = None
        self._sample_submission = None
        self._train_soundscapes_labels = None
        self._train = None

    @property
    def taxonomy(self):
        if self._taxonomy is None:
            self._taxonomy = pd.read_csv(self._taxonomy_path)

        return self._taxonomy

    @property
    def sample_submission(self):
        if self._sample_submission is None:
            self._sample_submission = pd.read_csv(self._sample_submission_path)

        return self._sample_submission

    @property
    def train_soundscapes_labels(self):
        if self._train_soundscapes_labels is None:
            self._train_soundscapes_labels = pd.read_csv(
                self._train_soundscapes_labels_path
            )

        return self._train_soundscapes_labels

    @property
    def train(self):
        if self._train is None:
            self._train = pd.read_csv(self._train_path)

        return self._train
```

# 2 - Visualization

## RUN

```python
data = BirdCLEFData(DATA_DIR)
```

### taxonomy

```python
data.taxonomy
```

```python
fig = px.histogram(data.taxonomy, x="class_name", title="Classes Histogram")
fig.show()
```

### train

```python
data.train.head(10)
```

```python
fig = px.histogram(
    data.train,
    x="common_name",
    color="class_name",
    title="Species Histogram",
    hover_data={"primary_label": True, "latitude": True},
    marginal="rug",
    height=500,
)
fig.show()
```

```python
train_summary = (
    data.train.groupby(["primary_label", "common_name", "class_name"])
    .agg(
        count=("primary_label", "count"),
        lat_mean=("latitude", "mean"),
        lat_std=("latitude", "std"),
        long_mean=("longitude", "mean"),
        long_std=("longitude", "std"),
    )
    .fillna({"lat_std": 0, "long_std": 0})
    .reset_index()
)
```

```python
train_summary
```

```python
fig = px.scatter(
    train_summary,
    x="common_name",
    y="lat_mean",
    error_y="lat_std",
    color="class_name",
    title="Latitude Distribution",
    height=500,
)
fig.show()
```

```python
fig = px.scatter(
    train_summary,
    x="common_name",
    y="long_mean",
    error_y="long_std",
    color="class_name",
    title="Longitude Distribution",
    height=500,
)
fig.show()
```
